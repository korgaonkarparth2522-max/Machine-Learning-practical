```python
# Task 1: Import Required Libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
# Task 2: Load Titanic Dataset

df = pd.read_csv('/content/Titanic-Dataset.csv')

print("Titanic dataset loaded successfully!")
```

```python
# Task 3: Display First 10 Rows

display(df.head(10))
```

```python
# Task 4: Display Last 5 Rows

display(df.tail(5))
```

```python
# Task 6: Display Dataset Information

df.info()
```

```python
# Task 7: Summary Statistics

display(df.describe())

print("\nAge Statistics:")
print("Mean Age:", df['Age'].mean())
print("Minimum Age:", df['Age'].min())
print("Maximum Age:", df['Age'].max())

print("\nFare Statistics:")
print("Mean Fare:", df['Fare'].mean())
print("Minimum Fare:", df['Fare'].min())
print("Maximum Fare:", df['Fare'].max())
```

```python
# Task 8: Check Missing Values

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
# Task 9: Fill Missing Age Values with Median

age_median = df['Age'].median()

print("Median Age:", age_median)

df['Age'] = df['Age'].fillna(age_median)

print("\nMissing Age values after filling:")
print(df['Age'].isnull().sum())
```

```python
# Task 10: Filter Surviving Passengers

survived_passengers = df[df['Survived'] == 1]

print("Number of surviving passengers:")
print(len(survived_passengers))

print("\nFirst 10 surviving passengers:")
display(survived_passengers.head(10))
```

```python
# Task 11: Convert Fare Column to NumPy Array

fare_array = df['Fare'].to_numpy()

print("First 10 Fare values:")
print(fare_array[:10])

print("\nType of Fare array:")
print(type(fare_array))
```

```python
# Task 12: Calculate Maximum and Average Fare

maximum_fare = np.max(fare_array)
average_fare = np.mean(fare_array)

print("Maximum Fare:", maximum_fare)
print("Average Fare:", average_fare)
```

```python
# Task 13: Histogram of Age

plt.figure(figsize=(8, 5))

plt.hist(
    df['Age'],
    bins=20
)

plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.title('Distribution of Passenger Age')

plt.show()
```

```python
# Task 14: Survival Count Bar Chart

survival_counts = df['Survived'].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    ['Did Not Survive', 'Survived'],
    survival_counts
)

plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.title('Titanic Survival Count')

plt.show()

print("Survival Counts:")
print(survival_counts)
```

```python
# Task 15: Age vs Fare Scatter Plot

plt.figure(figsize=(8, 5))

plt.scatter(
    df['Age'],
    df['Fare'],
    alpha=0.5
)

plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Age vs Fare')

plt.show()
```

The Titanic dataset was successfully loaded, inspected, cleaned, manipulated, and visualized using Python libraries in Google Colab. Surviving passengers were filtered, the Fare column was converted into a NumPy array, and the maximum and average Fare were calculated. Histograms, bar charts, and scatter plots were used to visualize the dataset and understand passenger age, survival counts, and the relationship between Age and Fare.
